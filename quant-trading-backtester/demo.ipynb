# Algorithmic Trading Backtester Demo
Run a quick momentum backtest on AAPL and inspect metrics.

```python
from backtester.engine import Backtester
from strategies.momentum import MomentumStrategy

backtester = Backtester()
backtester.add_strategy(MomentumStrategy())
metrics = backtester.run('AAPL', '1d', '2020-01-01', '2024-01-01')
metrics
```
